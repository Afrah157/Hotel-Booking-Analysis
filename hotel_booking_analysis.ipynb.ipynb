## 2. Clean the Data

### A. Missing value checking 

```python
df.isnull().sum()
```

### B. Missing value handling 

```python
df["children"] = df["children"].fillna(0)
df["country"] = df["country"].fillna("Unknown")
df["agent"] = df["agent"].fillna(0)
df["company"] = df["company"].fillna(0)
```

```python
df.isnull().sum()
```

### C. Duplicate checking

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

### D. Data type correction 

```python
df["children"] = df["children"].astype(int)
df["agent"] = df["agent"].astype(int)
df["company"] = df["company"].astype(int)
df["reservation_status_date"]=pd.to_datetime(df["reservation_status_date"])
```

```python
df.dtypes
```

### E. Removal of unnecessary columns if required

```python
df = df.drop(columns=["company"])
```

## 3. Create New Columns

### Create : Total_Guests

```python
df["Total_Guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)
df
```

### Create : Stay_Nights

```python
df["Stay_Nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)
df
```

### Create : Booking_Status

```python
df["Booking_Status"] = df["is_canceled"].map({
    1: "Cancelled",
    0: "Completed"
})
df
```

## 4. Basic Analysis

### A. Total Bookings

```python
total_bookings = len(df)
print("Total Bookings:", total_bookings)
```

### B. Total cancelled bookings

```python
total_cancelled = (df["is_canceled"] == 1).sum()
print("Total Cancelled Bookings:", total_cancelled)
```

### C. Cancellation Rate

```python
cancellation_rate = (total_cancelled / total_bookings) * 100
print("Cancellation Rate:", cancellation_rate, "%")
```

### D. Average Daily Rate

```python
average_daily_rate = df["adr"].mean()
print("Average Daily Rate:", average_daily_rate)
```

### E. Average Stay Duration

```python
average_stay = df["Stay_Nights"].mean()
print("Average Stay Duration:", average_stay)
```

### F. City Hotel bookings

```python
city_hotel_bookings = (df["hotel"] == "City Hotel").sum()
print("City Hotel Bookings:", city_hotel_bookings)
```

### G. Resort Hotel bookings

```python
resort_hotel_bookings = (df["hotel"] == "Resort Hotel").sum()
print("Resort Hotel Bookings:", resort_hotel_bookings)
```

## Export the Cleaned dataset
